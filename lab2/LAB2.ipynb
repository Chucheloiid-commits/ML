## 1. Первичный аудит данных

В начале работы были загружены два датасета: данные о ДТП и данные об автомобилях.

С помощью `head()` и `tail()` были выведены первые и последние строки таблиц. Также была определена их размерность:

- датасет ДТП — 4106 строк и 32 столбца;
- датасет автомобилей — 205 строк и 10 столбцов.

С помощью `info()` и `dtypes` были проверены типы данных. Числовые значения определились как числовые типы, а текстовые признаки — как `object`. В целом типы данных соответствуют содержимому таблиц.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

rta = pd.read_csv("test.csv")
auto = pd.read_csv("car_data.csv")

print("Размер датасета ДТП:", rta.shape)
print("Размер датасета автомобилей:", auto.shape)

print("\nДатасет ДТП")

print("\nПервые 5 строк:")
display(rta.head())

print("\nПоследние 5 строк:")
display(rta.tail())

print("\nИнформация о данных:")
rta.info()

print("\nТипы данных:")
display(rta.dtypes)


print("\nДатасет автомобилей")

print("\nПервые 5 строк:")
display(auto.head())

print("\nПоследние 5 строк:")
display(auto.tail())

print("\nИнформация о данных:")
auto.info()

print("\nТипы данных:")
display(auto.dtypes)
```

## 2. Статистический анализ и очистка данных

Для числовых признаков обоих датасетов были рассчитаны основные статистические показатели: среднее значение, стандартное отклонение, минимальное и максимальное значения, а также квартили. Для категориальных признаков определены количество уникальных значений и мода.

В датасете ДТП пропущенных значений и дубликатов не обнаружено, поэтому дополнительная очистка данных не потребовалась.

В датасете автомобилей пропущенных значений также не обнаружено. При этом была найдена одна дублирующаяся строка, которая была удалена. После очистки в обоих датасетах отсутствуют пропущенные значения и дубликаты.

```python
print("Числовые признаки ДТП:")
display(rta.describe())

print("Категориальные признаки ДТП:")
for column in rta.select_dtypes(include="str").columns:
    print(column, "— уникальных:", rta[column].nunique(),
          "| мода:", rta[column].mode()[0])

print("Пропуски в ДТП:")
display(rta.isnull().sum())

print("Дубликаты в ДТП:", rta.duplicated().sum())

# Заполняем пропуски
for column in rta.select_dtypes(include="str").columns:
    if rta[column].isnull().sum() > 0:
        rta[column] = rta[column].fillna(rta[column].mode()[0])

for column in rta.select_dtypes(include=np.number).columns:
    if rta[column].isnull().sum() > 0:
        rta[column] = rta[column].fillna(rta[column].median())

rta = rta.drop_duplicates()

print("После очистки:")
print("Пропуски:", rta.isnull().sum().sum())
print("Дубликаты:", rta.duplicated().sum())
```

```python
print("Числовые признаки автомобилей:")
display(auto.describe())

print("Категориальные признаки автомобилей:")
for column in auto.select_dtypes(include="str").columns:
    print(column, "— уникальных:", auto[column].nunique(),
          "| мода:", auto[column].mode()[0])

print("Пропуски в автомобилях:")
display(auto.isnull().sum())

print("Дубликаты в автомобилях:", auto.duplicated().sum())

# Заполняем пропуски
for column in auto.select_dtypes(include="str").columns:
    if auto[column].isnull().sum() > 0:
        auto[column] = auto[column].fillna(auto[column].mode()[0])

for column in auto.select_dtypes(include=np.number).columns:
    if auto[column].isnull().sum() > 0:
        auto[column] = auto[column].fillna(auto[column].median())

auto = auto.drop_duplicates()

print("После очистки:")
print("Пропуски:", auto.isnull().sum().sum())
print("Дубликаты:", auto.duplicated().sum())
```

### 3.1. Гистограмма количества транспортных средств

Гистограмма показывает, как часто встречаются различные значения признака. В данном случае исследуется количество транспортных средств, участвовавших в ДТП. С помощью `histplot` также добавлен график плотности (`kde=True`), чтобы лучше увидеть форму распределения данных.

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.histplot(data=rta, x="Number_of_vehicles_involved", bins=6, kde=True)

plt.title("Распределение количества транспортных средств")
plt.xlabel("Количество транспортных средств")
plt.ylabel("Количество ДТП")
plt.show()
```

### 3.2. График плотности количества пострадавших

График плотности показывает форму распределения значений без отображения отдельных столбцов. В данном случае рассматривается количество пострадавших в ДТП. По графику можно определить, в области каких значений наблюдается наибольшая концентрация данных.

```python
sns.kdeplot(data=rta, x="Number_of_casualties", fill=True)

plt.title("Плотность распределения количества пострадавших")
plt.xlabel("Количество пострадавших")
plt.ylabel("Плотность")
plt.show()
```

### 4.1. Матрица корреляции

Корреляция показывает, насколько два числовых признака связаны между собой. Коэффициент корреляции принимает значения от -1 до 1. Значение, близкое к 1, означает сильную положительную связь, а значение, близкое к -1 — сильную отрицательную связь. Значение около 0 говорит об отсутствии выраженной линейной зависимости.

Сначала рассчитываем матрицу корреляции для всех числовых признаков датасета ДТП.

```python
corr_rta = rta.select_dtypes(include=np.number).corr()

display(corr_rta)
```

### 4.2. Тепловая карта корреляции

Для более наглядного представления матрицы корреляции построим тепловую карту. Цвет ячейки показывает направление и силу связи между признаками, а значение коэффициента выводится непосредственно на графике.

```python
plt.figure(figsize=(8, 6))

sns.heatmap(corr_rta, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Матрица корреляции числовых признаков ДТП")
plt.show()
```

### 4.3. Диаграмма рассеяния

После построения матрицы корреляции выберем пару числовых признаков с наиболее заметной линейной зависимостью и построим диаграмму рассеяния. Каждая точка на графике соответствует одному наблюдению. Такой график позволяет визуально оценить характер связи между двумя признаками.

```python
sns.scatterplot(
    data=rta,
    x="Number_of_vehicles_involved",
    y="Number_of_casualties"
)

plt.title("Зависимость количества пострадавших от количества транспортных средств")
plt.xlabel("Количество транспортных средств")
plt.ylabel("Количество пострадавших")
plt.show()
```

### 4.4. Матрица корреляции для датасета автомобилей

Для датасета автомобилей рассчитаем матрицу корреляции между числовыми признаками. Коэффициент корреляции показывает направление и силу линейной зависимости между признаками. Значение коэффициента находится в диапазоне от -1 до 1.

```python
corr_auto = auto.select_dtypes(include=np.number).corr()

display(corr_auto)
```

### 4.5. Тепловая карта корреляции

Для наглядного представления матрицы корреляции построим тепловую карту. В каждой ячейке будет отображаться коэффициент корреляции, а цвет будет показывать направление и силу зависимости между признаками.

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    corr_auto,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Матрица корреляции числовых признаков автомобилей")
plt.show()
```

### 4.6. Диаграмма рассеяния

По матрице корреляции наиболее сильная линейная зависимость наблюдается между расходом топлива в городе (`city-mpg`) и расходом топлива на трассе (`highway-mpg`). Коэффициент корреляции равен 0,971, что указывает на очень сильную положительную линейную связь между признаками.

Для визуальной проверки этой зависимости построим диаграмму рассеяния.

```python
sns.scatterplot(
    data=auto,
    x="city-mpg",
    y="highway-mpg"
)

plt.title("Зависимость расхода топлива на трассе от расхода топлива в городе")
plt.xlabel("Расход топлива в городе (city-mpg)")
plt.ylabel("Расход топлива на трассе (highway-mpg)")
plt.show()
```
